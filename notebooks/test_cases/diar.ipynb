```python
import pandas as pd

print("--- 1. LOADING CLINICAL DIARRHEA DATA FROM STATA FILE ---")
# Ensure your file name matches exactly what is in E:\BTP>
file_path = 'dia_sample.dta' 

# Read the Stata file
df = pd.read_stata(file_path)

# Step A: Use 'clin2' as the Date of Onset
date_column = 'clin2'

# Step B: Convert to datetime safely
df['Parsed_Date'] = pd.to_datetime(df[date_column], errors='coerce')

# Step C: Drop rows where Date or the Cholera results ('v_chol') are missing
clean_df = df.dropna(subset=['Parsed_Date', 'v_chol']).copy()

# Step D: Standardize the date text format for easy filtering later
clean_df['Formatted_Date'] = clean_df['Parsed_Date'].dt.strftime('%Y-%m-%d')

# Step E: Print out the statistical summary of your dataset
print("\n============================ DATA METRICS ============================")
print(f"Total Usable Patient Records (with Date & Cholera Results): {len(clean_df)}")
print("----------------------------------------------------------------------")
print("Cholera (v_chol) Test Results Breakdown:")
print(clean_df['v_chol'].value_counts())
print("======================================================================\n")

# Step F: Extract timeline parameters for your Google Trends payload
start_date = clean_df['Formatted_Date'].min()
end_date = clean_df['Formatted_Date'].max()

print("Timeline parameters detected for your next script:")
print(f"Start Date: {start_date}")
print(f"End Date: {end_date}")
```
