```python
import pandas as pd

FILE_PATH = r"E:\BTP\Data\Raw_data\Yemen Cholera Outbreak Epidemiology Data - Data_Governorate_Level.csv"

df = pd.read_csv(FILE_PATH)

unique_names = sorted(df["Governorate"].dropna().unique())

print(f"Total unique governorate names: {len(unique_names)}")
print("\nUnique names:")

for i, name in enumerate(unique_names, start=1):
    print(f"{i}. {name!r}")
```

```python
import pandas as pd

FILE_PATH = r"E:\BTP\Data\Raw_data\Yemen Cholera Outbreak Epidemiology Data - Data_Governorate_Level.csv"

df = pd.read_csv(FILE_PATH)

# Count every exact value in the Governorate column
name_counts = df["Governorate"].value_counts(dropna=False)

print(f"Total rows: {len(df)}")
print(f"Total unique Governorate values: {df['Governorate'].nunique(dropna=False)}")

print("\nGovernorate values and their frequencies:")
print("-" * 60)

for name, count in name_counts.items():
    print(f"{name!r:<35} -> {count} rows")
```
