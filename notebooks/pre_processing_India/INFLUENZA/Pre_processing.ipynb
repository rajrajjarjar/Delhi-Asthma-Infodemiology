```python
"""
Script 1: Count how many rows have an actual (non-empty) value for each
virus type in the India FluNet file, sorted from most to least reported.

"Actual value" = the cell is not blank/NaN. A blank means that virus
wasn't tested for / reported that week, not that the count was zero.
"""

import pandas as pd

# ---- fill these in before running ----
raw_path = "E:/Projects/BTP/Data/Clean/INDIA/Influenza/india_flu.csv"       # path to input CSV (e.g. india_flu.csv)
output_path = "E:/Projects/BTP/Data/Clean/INDIA/Influenza/selecting_virus.csv"    # path to save the results CSV
# ----------------------------------------

# every virus/subtype column present in the FluNet schema
VIRUS_COLUMNS = [
    "AH1N12009", "AH1", "AH3", "AH5", "AH7N9",
    "ANOTSUBTYPED", "ANOTSUBTYPABLE", "AOTHER_SUBTYPE", "INF_A",
    "BVIC_2DEL", "BVIC_3DEL", "BVIC_NODEL", "BVIC_DELUNK",
    "BYAM", "BNOTDETERMINED", "INF_B", "INF_ALL",
    "ADENO", "BOCA", "HUMAN_CORONA", "METAPNEUMO",
    "PARAINFLUENZA", "RHINO", "RSV", "OTHERRESPVIRUS",
]


def count_reported_rows(df: pd.DataFrame) -> pd.DataFrame:
    """
    For each virus column, count how many rows are NOT empty (i.e. a
    value was actually reported that week), then sort descending so
    the most-frequently-reported virus is on top.
    """
    counts = {}
    for col in VIRUS_COLUMNS:
        if col in df.columns:
            counts[col] = df[col].notna().sum()

    result = (
        pd.DataFrame(list(counts.items()), columns=["virus_type", "rows_with_data"])
        .sort_values("rows_with_data", ascending=False)
        .reset_index(drop=True)
    )
    return result


def main():
    # load the raw data, skipping any malformed lines instead of crashing
    df = pd.read_csv(raw_path)

    result = count_reported_rows(df)

    # save output and also print to console for a quick look
    result.to_csv(output_path, index=False)
    print(result.to_string(index=False))


if __name__ == "__main__":
    main()
```
